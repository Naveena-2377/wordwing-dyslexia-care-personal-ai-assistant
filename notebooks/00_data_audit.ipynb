# Data audit

- Count samples per class in each dataset
- Check image sizes, audio durations, sample rates
- Flag corrupt/unreadable files before you waste a training run

```python
import sys; sys.path.append('../src')
from pathlib import Path
from PIL import Image

root = Path("../data/raw/m1_dyslexia_handwriting")
for split in ["Train", "Test"]:
    for cls in ["Normal", "Reversal", "Corrected"]:
        folder = root / split / cls
        n = len(list(folder.glob("*.png")))
        print(f"{split}/{cls}: {n} images")
```

```python
sample = next((root / "Train" / "Normal").glob("*.png"))
img = Image.open(sample)
print(img.size, img.mode)
```
